```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load data
df = pd.read_csv('Dataset/Q06_museum_visitors.csv')
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])                          
df['visitor_count'].fillna(df['visitor_count'].median(), inplace=True)  
df['weather'].fillna('Unknown', inplace=True)                    
df['ticket_type'].fillna('Unknown', inplace=True)                


print(df.isnull().sum())
df.dtypes
```

```python
df['date']
```

```python
# 3a. Visitors by weekday
df['weekday'] = df['date'].dt.day_name()
weekday_order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']

df.groupby('weekday')['visitor_count'].mean().reindex(weekday_order).plot(kind='bar', title='Avg Visitors by Weekday', color='steelblue')
plt.ylabel('Avg Visitors'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

```python

df.groupby('weather')['visitor_count'].mean().sort_values().plot(kind='bar', title='Avg Visitors by Weather', color='coral')
plt.ylabel('Avg Visitors'); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```

```python

df.groupby('special_exhibition')['visitor_count'].mean().sort_values().plot(kind='bar', title='Avg Visitors by Exhibition', color='mediumseagreen')
plt.ylabel('Avg Visitors'); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```

```python

df['is_weekend'] = df['date'].dt.dayofweek >= 5

footfall_per_day = df.groupby('special_exhibition').apply(
    lambda x: x['visitor_count'].sum() / x['date'].nunique()
).rename('footfall_per_exhibition_day').reset_index()

print(footfall_per_day)
footfall_per_day.plot(x='special_exhibition', y='footfall_per_exhibition_day', kind='bar',
                      title='Footfall per Exhibition Day', color='mediumpurple', legend=False)
plt.ylabel('Visitors / Day'); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```

```python
df.groupby('is_weekend')['visitor_count'].mean().rename(index={False:'Weekday', True:'Weekend'}).plot(
    kind='bar', title='Avg Visitors: Weekday vs Weekend', color=['steelblue','coral'])
plt.ylabel('Avg Visitors'); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```
